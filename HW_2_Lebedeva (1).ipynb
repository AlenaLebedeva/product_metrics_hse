### Warm up

#### 1. Game dev — 2 балла
У вас есть данные о событиях в мобильной игре. Рассчитайте и визуализируйте:
- User Stickiness — отношение DAU к MAU (MAU рассчитывайте скользящим окном 30 дней для каждой даты только для полного окна)
- ARPDAU
- Paying Share — доля платящих пользователей
- Sessions per User — среднее число сессий на пользователя
- Daily Play Time — среднее время, проводимое пользователем в игре в день (рассчитываем как общее время в приложении)
- Avg Attempts per Session — среднее число попыток на сессию

*Смотрим на подневную динамику

```python
df_game = pd.read_csv('hw_2_gamedev.csv', index_col=0)
df_game.head()
```

Описание данных (уникальный ключ session_id)

- session_id — уникальный идентификатор сессии
- user — уникальный идентификатор юзера
- date — дата
- revenue — суммарная выручка на сессию
- attempts — число попыток на сессию
- session_length_sec — длина сессии в секундах

```python
df_game['date'] = pd.to_datetime(df_game['date'])

# Расчет User Stickiness

df = df_game.groupby('date')['user'].nunique().reset_index()
df.columns = ['date', 'dau']

mau = []
dates = df['date']
for i in range(len(dates)):
    if i < 29:
        mau.append(None)
    else:
        mau.append(df_game[(df_game['date'] >= dates[i - 29]) & (df_game['date'] <= dates[i])]['user'].nunique())

df['mau'] = mau

df['user_stickiness'] = df['dau'] / df['mau']
df

# Визуализация

plt.figure(figsize=(18, 6))
plt.plot(df['date'], df['user_stickiness'], marker = 'o')
plt.title('User Stickiness')
plt.ylabel('User Stickiness')
plt.xlabel('Дата')
plt.grid(True)
plt.show()
```

```python
# Расчет ARPDAU

df = df_game.groupby('date').agg(users = ('user', 'nunique'), revenue = ('revenue', 'sum')).reset_index()
df.columns = ['date', 'dau', 'revenue']
df['arpdau'] = df['revenue'] / df['dau']
df

# Визуализация

plt.figure(figsize=(18, 6))
plt.plot(df['date'], df['arpdau'], marker = 'o')
plt.title('Средний подневный доход на 1 активного пользователя')
plt.ylabel('ARPDAU')
plt.xlabel('Дата')
plt.grid(True)
plt.show()
```

```python
# Расчет Paying Share

df_rev = df_game.query('revenue > 0')
df_rev = df_rev.groupby('date')['user'].nunique().reset_index()
df_rev.columns = ['date', 'paying']
df = df_game.groupby('date')['user'].nunique().reset_index()
df.columns = ['date', 'dau']
df = pd.merge(df, df_rev, how='left', on='date')
df['percentage_of_paying'] = df['paying'] / df['dau']

# Визуализация

plt.figure(figsize=(18, 6))
plt.plot(df['date'], df['percentage_of_paying'], marker = 'o')
plt.title('Подневые доли платящих пользователей')
plt.ylabel('Доля платящих')
plt.xlabel('Дата')
plt.grid(True)
plt.show()
```

```python
# Расчет Sessions per User

df = df_game.groupby(['date', 'user'])['session_id'].nunique().reset_index()
df.columns = ['date', 'user', 'sessions']
df = df.groupby('date').agg(users = ('user', 'nunique'), sessions = ('sessions', 'sum')).reset_index()
df.columns = ['date', 'users', 'sessions']
df['avg_sessions'] = df['sessions'] / df['users']

# Визуализация

plt.figure(figsize=(18, 6))
plt.plot(df['date'], df['avg_sessions'], marker = 'o')
plt.title('Динамика среднего числа сессий на пользователя')
plt.ylabel('Число сессий')
plt.xlabel('Дата')
plt.grid(True)
plt.show()
```

```python
# Расчет Daily Play Time

df = df_game.groupby('date').agg(users = ('user', 'nunique'), duration = ('session_length_sec', 'sum')).reset_index()
df['avg_duration'] = df['duration'] / df['users']

# Визуализация

plt.figure(figsize=(18, 6))
plt.plot(df['date'], df['avg_duration'], marker = 'o')
plt.title('Динамика средней продолжительности времени в игре на пользователя')
plt.ylabel('Время, сек')
plt.xlabel('Дата')
plt.grid(True)
plt.show()
```

```python
# Расчет Avg Attempts per Session

df = df_game.groupby('date').agg(sessions = ('session_id', 'nunique'), attempts = ('attempts', 'sum')).reset_index()
df['avg_attempts'] = df['attempts'] / df['sessions']

# Визуализация

plt.figure(figsize=(18, 6))
plt.plot(df['date'], df['avg_attempts'], marker = 'o')
plt.title('Динамика среднего числа попыток на сессию')
plt.ylabel('Число попыток')
plt.xlabel('Дата')
plt.grid(True)
plt.show()
```

#### 2. Ride hailing  — 2 балла
У вас есть данные из приложения такси. Рассчитайте и визуализируйте:
- Ride Completion Rate — CR из заказа в завершенную поездку
- Acceptance Rate — доля заказов, принятых водителем (считаем, что приняты все заказы, кроме заказов со статусом 'No Driver Found')
- Cancellation Rate — доля отмененных заказов
- Rides per Driver — среднее число поездок на водителя
- Avg Ride Check — средний чек поездки
- Отношение RTA к ETA (что измеряет эта метрика?)

*Смотрим на подневную динамику

```python
df_rh = pd.read_csv('hw_2_taxi.csv')
```

```python
df_rh.head()
```

Описание данных (уникальный ключ order_uid)

- date — дата создания заказа
- time — время создания заказа
- order_uid — уникальный идентификатор заказа
- order_status — статус заказа
- passenger_id — уникальный идентификатор пассажира
- vehicle_type — тип транспорта
- pickup_location — место отправления
- drop_location — место назначения
- ETA — прогнозное время прибытия водителя на место отправления (estimated time of arrival)
- RTA — реальное время прибытия водителя на место отправления (real time of arrival)
- driver_id — уникальный идентификатор водителя
- cancelled_rides_by_customer — флаг отмены поездки пассажиром
- reason_for_cancelling_by_customer — причина отмены поездки пассажиром
- cancelled_rides_by_driver — флаг отмены поездки водителем
- driver_cancellation_reason — причина отмены поездки водителем
- incomplete_rides — флаг незавершенной поездки
- incomplete_rides_reason — причина незавершенной поездки
- fare — цена поездки
- ride_distance – расстояние поездки
- driver_ratings — рейтинг водителя
- customer_rating — рейтинг пассажира
- payment_method – способ оплаты

```python
df_rh['date'] = pd.to_datetime(df_rh['date'])

# Расчет Ride Completion Rate

df = df_rh.copy()
df['completed'] = df['order_status'].apply(lambda x: 1 if x == 'Completed'
                                           else 0)
df = df.groupby('date').agg(orders = ('order_uid', 'nunique'), completed = ('completed', 'sum')).reset_index()
df['cv_compl'] = df['completed'] / df['orders'] * 100

# Визуализация

plt.figure(figsize=(18, 6))
plt.plot(df['date'], df['cv_compl'], marker = 'o')
plt.title('Динамика конверсии из заказа в совершенную поездку')
plt.ylabel('Конверсия, %')
plt.xlabel('Дата')
plt.grid(True)
plt.show()
```

```python
# Расчет Acceptance Rate

df = df_rh.copy()
df_driver = df.query('order_status != "No Driver Found"')
df_driver = df_driver.groupby('date')['order_uid'].nunique().reset_index()
df_driver.columns = ['date', 'driver_found']
df = df.groupby('date')['order_uid'].nunique().reset_index()
df.columns = ['date', 'orders']
df = pd.merge(df, df_driver, how='left', on='date')
df['acceptance_rate'] = df['driver_found'] / df['orders']

# Визуализация

plt.figure(figsize=(18, 6))
plt.plot(df['date'], df['acceptance_rate'], marker = 'o')
plt.title('Динамика доли заказов, принятых водителем')
plt.ylabel('Доля принятых заказов')
plt.xlabel('Дата')
plt.grid(True)
plt.show()
```

```python
# Расчет Cancellation Rate

df = df_rh.copy()
df['cancelled'] = df['order_status'].apply(lambda x: 1 if x == 'Cancelled by Driver'
                                           else 1 if x == 'Cancelled by Customer'
                                           else 0)
df = df.groupby('date').agg(orders = ('order_uid', 'nunique'), cancelled = ('cancelled', 'sum')).reset_index()
df['cancellation_rate'] = df['cancelled'] / df['orders']

# Визуализация

plt.figure(figsize=(18, 6))
plt.plot(df['date'], df['cancellation_rate'], marker = 'o')
plt.title('Динамика доли отмененных заказов')
plt.ylabel('Доля отмененных заказов')
plt.xlabel('Дата')
plt.grid(True)
plt.show()
```

```python
# Расчет Rides per Driver

df = df_rh.query('order_status != "No Driver Found"')
df = df.groupby('date').agg(orders = ('order_uid', 'nunique'), drivers = ('driver_id', 'nunique')).reset_index()
df['rides_per_driver'] = df['orders'] / df['drivers']

# Визуализация

plt.figure(figsize=(18, 6))
plt.plot(df['date'], df['rides_per_driver'], marker = 'o')
plt.title('Динамика среднего числа поездок на водителя')
plt.ylabel('Число поездок')
plt.xlabel('Дата')
plt.grid(True)
plt.show()
```

```python
# Расчет Avg Ride Check

df = df_rh.query('fare > 0')
df = df.groupby('date').agg(orders = ('order_uid', 'nunique'), revenue = ('fare', 'sum')).reset_index()
df['avg_ride_check'] = df['revenue'] / df['orders']

# Визуализация

plt.figure(figsize=(18, 6))
plt.plot(df['date'], df['avg_ride_check'], marker = 'o')
plt.title('Динамика среднего чека поездки')
plt.ylabel('Цена поездки')
plt.xlabel('Дата')
plt.grid(True)
plt.show()
```

```python
# Расчет Отношение RTA к ETA

df = df_rh[df_rh['RTA'].notna()]
df['rta_to_eta'] = df['RTA'] / df['ETA']

df = df.groupby('date')['rta_to_eta'].mean().reset_index()

# Визуализация

plt.figure(figsize=(18, 6))
plt.plot(df['date'], df['rta_to_eta'], marker = 'o')
plt.title('Динамика отношения RTA к ETA')
plt.ylabel('Отношение RTA к ETA')
plt.xlabel('Дата')
plt.grid(True)
plt.show()
```

***Что измеряет метрика "Отношение RTA к ETA"?***

Эта метрика показывает, насколько фактическое время прибытия водителя соответствует прогнозируемому.
Наилучший результат достигается при значении <= 1, это означает, что водитель выполнил поездку вовремя либо раньше спрогнозированного времени.

```python
df_all_info = df_music_logs.merge(df_music_subs, on = 'uid', how = 'left')
df_all_info['listened_seconds'] = df_all_info['played_ratio_pct'] / 100 * df_all_info['track_length_seconds']
```

```python
# Среднее время прослушивания на пользователя

df_all_info['date'] = df_all_info['datetime'].dt.date
dates = sorted(df_all_info['date'].unique())
mau = []
wau = []

for current_date in dates:
    start_date = current_date - pd.Timedelta(days=29)
    users_window = df_all_info[(df_all_info['date'] >= start_date) & (df_all_info['date'] <= current_date)]
    mau.append(users_window['listened_seconds'].sum() / users_window['uid'].nunique())

for current_date in dates:
    start_date = current_date - pd.Timedelta(days=6)
    users_window = df_all_info[(df_all_info['date'] >= start_date) & (df_all_info['date'] <= current_date)]
    wau.append(users_window['listened_seconds'].sum() / users_window['uid'].nunique())

df_all_info_mau_series = pd.Series(mau, index=dates)
df_all_info_wau_series = pd.Series(df_all_info_mau_series = pd.Series(mau, index=dates), index=dates)

plt.figure(figsize=(18, 6))

plt.plot(df_all_info_wau_series.index, df_all_info_wau_series.values, marker='o', label='Среднее время прослушиваний по неделям')
plt.plot(df_all_info_mau_series.index, df_all_info_mau_series.values, marker='o', label='Среднее время прослушиваний по месяцам')

plt.title('Динамика среднего времени прослушивания на 1 пользователя среди всех пользователей по неделям и месяцам')
plt.xlabel('Дата')
plt.ylabel('Время прослушивания')
plt.legend()
plt.grid(True)

plt.show()
```
