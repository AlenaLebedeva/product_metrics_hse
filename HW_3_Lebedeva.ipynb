#### 2. Рассчитайте количество дней, необходимое для детектирования заявленных в гипотезе эффектов (target и guardrails-метрики) — 2 балла
Здесь и далее используйте для расчетов значение alpha = 0.01 (делаем поправку на принятие решения по набору метрик) и beta = 0.2. А также двусторонний критерий — мы заранее не знаем, куда двинутся метрики, поэтому хотим учесть и противоположные эффекты.  

Для метрики "доля пользователей с контактом по объявлению из любой категории" также используйте эффект в 3%.

```python
def filter_users_by_event(df, event_name, category=None, source=None):
    df_filtered = df[df['event'] == event_name]
    if category is not None:
        df_filtered = df_filtered[df_filtered['category'].isin(category)]
    if source is not None:
        df_filtered = df_filtered[df_filtered['source'] == source]
    return set(df_filtered['user_id'].unique())

def calc_conversion(df, all_users_set, event_name, category=None, source=None):
    users_with_event = filter_users_by_event(df, event_name, category, source)
    conv = len(users_with_event.intersection(all_users_set)) / len(all_users_set)
    return conv

def calc_sample_size(p0, relative_effect, alpha=0.01, beta=0.2):
    p1 = p0 * (1 + relative_effect)
    p_bar = (p0 + p1) / 2
    q_bar = 1 - p_bar
    z_alpha = norm.ppf(1 - alpha/2)
    z_beta = norm.ppf(1 - beta)
    n = ((z_alpha * math.sqrt(2 * p_bar * q_bar) + z_beta * math.sqrt(p0 * (1 - p0) + p1 * (1 - p1)))**2) / (p1 - p0)**2
    return math.ceil(n)

all_users = set(df_logs['user_id'].unique())

conv_target = calc_conversion(df_logs, all_users, 'contact_submit', category=['newbuilds'])
conv_rent = calc_conversion(df_logs, all_users, 'contact_submit', category=['rent_daily', 'rent_long'])
conv_secondary = calc_conversion(df_logs, all_users, 'contact_submit', category=['secondary'])
conv_any = calc_conversion(df_logs, all_users, 'contact_submit', category=None)

size_target = calc_sample_size(conv_target, 0.15)
size_rent = calc_sample_size(conv_rent, -0.03)
size_secondary = calc_sample_size(conv_secondary, -0.03)
size_any = calc_sample_size(conv_any, -0.03)

df_logs['timestamp'] = pd.to_datetime(df_logs['timestamp'])
df_logs['date'] = df_logs['timestamp'].dt.date
n_users = df_logs.groupby('date')['user_id'].nunique().reset_index()
n_users_avg = n_users['user_id'].mean()
n_target = size_target / n_users_avg
n_rent = size_rent / n_users_avg
n_secondary = size_secondary / n_users_avg
n_any = size_any / n_users_avg

print(math.ceil(min(n_target, n_secondary, n_rent, n_any)))
```

#### 3. Рассчитайте MDE для информативных метрик — 2 балла
В реальности эффекты по информативным критериям, которые мы также хотим включать в пресет метрик, нужно сравнивать на достижимость на основе прошлых экспериментов. Как правило, проделав это упражнение один раз, вы просто сохраняете достижимый трешхолд в базе знаний команды и принимаете решение о включении их в базовый набор. Проверка эффектов для каждого теста не требуется. Однако, стоит делать периодический пересмотр зафиксированных достижимых эффектов, например раз в полгода / год, или при сильных изменениях продукта / рынка.

В данном задании мы с вами не будем опираться на историю тестов, просто потренируемся делать обратный расчет — размера эффекта от заданной длительности (14 дней).

*Для числителя используйте только ивенты с источником main_nav*

```python
def calc_mde(p0, n, alpha=0.01, beta=0.2):

    z_alpha = norm.ppf(1 - alpha/2)
    z_beta = norm.ppf(1 - beta)
    mde = (z_alpha + z_beta) * math.sqrt(2 * p0 * (1 - p0) / n)
    return mde

df_logs_main = df_logs[df_logs['source'] == 'main_nav']

n_users = (df_logs_main.groupby('date')['user_id'].nunique().reset_index())['user_id'].mean()

n = n_users * 14 / 2

all_users_main = set(df_logs_main['user_id'].unique())

conv_target_main = calc_conversion(df_logs_main, all_users, 'list_view', category=['newbuilds'], source='main_nav')
conv_rent_main = calc_conversion(df_logs_main, all_users, 'list_view', category=['rent_daily', 'rent_long'], source='main_nav')
conv_secondary_main = calc_conversion(df_logs_main, all_users, 'list_view', category=['secondary'], source='main_nav')
conv_any_main = calc_conversion(df_logs_main, all_users, 'list_view', category=None, source='main_nav')

mde_target = calc_mde(conv_target_main, n) / conv_target_main * 100
mde_rent = calc_mde(conv_rent_main, n) / conv_rent_main * 100
mde_secondary = calc_mde(conv_secondary_main, n) / conv_secondary_main * 100
mde_any = calc_mde(conv_any_main, n) / conv_any_main * 100

print(round(max(mde_target, mde_rent, mde_secondary, mde_any), 2))
```

```python
df_view = df_exps[['exp_id', 'target_contact_effect_rel', 'target_contact_p_value', 'proxy_card_effect_rel', 'proxy_card_p_value']].copy()
df_add = df_exps[['exp_id', 'target_contact_effect_rel', 'target_contact_p_value', 'proxy_fav_effect_rel', 'proxy_fav_p_value']].copy()
df_form = df_exps[['exp_id', 'target_contact_effect_rel', 'target_contact_p_value', 'proxy_form_effect_rel', 'proxy_form_p_value']].copy()

df_view['target_significant'] = df_view['target_contact_p_value'] < 0.05
df_add['target_significant'] = df_add['target_contact_p_value'] < 0.05
df_form['target_significant'] = df_form['target_contact_p_value'] < 0.05
df_view['proxy_significant'] = df_view['proxy_card_p_value'] < 0.05
df_add['proxy_significant'] = df_add['proxy_fav_p_value'] < 0.05
df_form['proxy_significant'] = df_form['proxy_form_p_value'] < 0.05

def count_binary_sensitivity(df):
  a = df[df['target_significant'] == True]['exp_id'].nunique()
  b = df[(df['target_significant'] == True) & (df['proxy_significant'] == True)]['exp_id'].nunique()
  return b / a

bin_sens_view = count_binary_sensitivity(df_view)
bin_sens_add = count_binary_sensitivity(df_add)
bin_sens_form = count_binary_sensitivity(df_form)

mse_view = ((df_view['target_contact_effect_rel'] - df_view['proxy_card_effect_rel']) ** 2).mean()
mse_add = ((df_add['target_contact_effect_rel'] - df_add['proxy_fav_effect_rel']) ** 2).mean()
mse_form = ((df_form['target_contact_effect_rel'] - df_form['proxy_form_effect_rel']) ** 2).mean()

corr_view, p_value_view = pearsonr(df_view['proxy_card_effect_rel'], df_view['target_contact_effect_rel'])
corr_add, p_value_add = pearsonr(df_add['proxy_fav_effect_rel'], df_add['target_contact_effect_rel'])
corr_form, p_value_form = pearsonr(df_form['proxy_form_effect_rel'], df_form['target_contact_effect_rel'])

print(round(max(corr_view, corr_add, corr_form), 2))
```

```python
df_view['target_sign'] = df_view['target_contact_effect_rel'] > 0
df_add['target_sign'] = df_add['target_contact_effect_rel'] > 0
df_form['target_sign'] = df_form['target_contact_effect_rel'] > 0
df_view['proxy_sign'] = df_view['proxy_card_effect_rel'] > 0
df_add['proxy_sign'] = df_add['proxy_fav_effect_rel'] > 0
df_form['proxy_sign'] = df_form['proxy_form_effect_rel'] > 0

def count_binary_sensitivity(df):

  df = df[df['target_significant'] == True]
  a = df['exp_id'].nunique()
  b = df[(df['proxy_significant'] == True) & (df['target_sign'] == df['proxy_sign'])]['exp_id'].nunique()
  c = df[(df['proxy_significant'] == True) & (df['target_sign'] != df['proxy_sign'])]['exp_id'].nunique()
  return (b - c) / a

score_view = count_binary_sensitivity(df_view)
score_add = count_binary_sensitivity(df_add)
score_form = count_binary_sensitivity(df_form)

confusion_view = pd.crosstab(df_view['proxy_sign'], df_view['target_sign'],
                        rownames=['proxy_sign'], colnames=['target_sign'])

confusion_add = pd.crosstab(df_add['proxy_sign'], df_add['target_sign'],
                        rownames=['proxy_sign'], colnames=['target_sign'])

confusion_form = pd.crosstab(df_form['proxy_sign'], df_form['target_sign'],
                        rownames=['proxy_sign'], colnames=['target_sign'])

print('Пользователи с просмотром карточки объявления в Новостройках')
print('Proxy score = ', score_view)
print('Матрица согласия-несогласия знаков:\n', confusion_view)

print('Пользователи, добавившие в избранное объявление в Новостройках')
print('Proxy score = ', score_add)
print('Матрица согласия-несогласия знаков:\n', confusion_add)

print('Пользователи, открывшие форму заявки в Новостройках')
print('Proxy score = ', score_form)
print('Матрица согласия-несогласия знаков:\n', confusion_form)
```
